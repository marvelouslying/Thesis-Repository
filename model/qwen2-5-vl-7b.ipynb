# 数据下载(从数据下载到生成csv的过程)

```python
# 导入所需的库
from modelscope.msdatasets import MsDataset
import os
import pandas as pd

MAX_DATA_NUMBER = 500

# 检查目录是否已存在
if not os.path.exists('coco_2014_caption'):
    # 从modelscope下载COCO 2014图像描述数据集
    ds =  MsDataset.load('modelscope/coco_2014_caption', subset_name='coco_2014_caption', split='train')
    print(len(ds))
    # 设置处理的图片数量上限
    total = min(MAX_DATA_NUMBER, len(ds))

    # 创建保存图片的目录
    os.makedirs('coco_2014_caption', exist_ok=True)

    # 初始化存储图片路径和描述的列表
    image_paths = []
    captions = []

    for i in range(total):
        # 获取每个样本的信息
        item = ds[i]
        image_id = item['image_id']
        caption = item['caption']
        image = item['image']

        # 保存图片并记录路径
        image_path = os.path.abspath(f'coco_2014_caption/{image_id}.jpg')
        image.save(image_path)

        # 将路径和描述添加到列表中
        image_paths.append(image_path)
        captions.append(caption)

        # 每处理50张图片打印一次进度
        if (i + 1) % 50 == 0:
            print(f'Processing {i+1}/{total} images ({(i+1)/total*100:.1f}%)')

    # 将图片路径和描述保存为CSV文件
    df = pd.DataFrame({
        'image_path': image_paths,
        'caption': captions
    })

    # 将数据保存为CSV文件
    df.to_csv('./coco-2024-dataset.csv', index=False)


else:
    print('coco_2014_caption目录已存在,跳过数据处理步骤')
```

# 模型微调(Lora微调)

```python
# torch.cuda.empty_cache()
```

```python
import torch
from datasets import Dataset
from modelscope import snapshot_download, AutoTokenizer
from swanlab.integration.transformers import SwanLabCallback
from qwen_vl_utils import process_vision_info
from peft import LoraConfig, TaskType, get_peft_model, PeftModel
from transformers import (
    TrainingArguments,
    Trainer,
    DataCollatorForSeq2Seq,
    Qwen2VLForConditionalGeneration,
    AutoProcessor,
)
import swanlab
import json
# 避免并行警告
import os
os.environ["TOKENIZERS_PARALLELISM"] = "false"
# PYTORCH_CUDA_ALLOC_CONF=expandable_segments:True # 有助于PyTorch的内存分配器更好地管理碎片

def process_func(example):
    """
    将数据集进行预处理
    """
    MAX_LENGTH = 8192
    input_ids, attention_mask, labels = [], [], []
    conversation = example["conversations"]
    input_content = conversation[0]["value"]
    output_content = conversation[1]["value"]
    file_path = input_content.split("<|vision_start|>")[1].split("<|vision_end|>")[0]  # 获取图像路径
    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": f"{file_path}",
                    "resized_height": 280,
                    "resized_width": 280,
                },
                {"type": "text", "text": "COCO Yes:"},
            ],
        }
    ]
    text = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )  # 获取文本
    image_inputs, video_inputs = process_vision_info(messages)  # 获取数据数据（预处理过）
    inputs = processor(
        text=[text],
        images=image_inputs,
        videos=video_inputs,
        padding=True,
        return_tensors="pt",
    )
    inputs = {key: value.tolist() for key, value in inputs.items()} #tensor -> list,为了方便拼接
    instruction = inputs

    response = tokenizer(f"{output_content}", add_special_tokens=False)


    input_ids = (
            instruction["input_ids"][0] + response["input_ids"] + [tokenizer.pad_token_id]
    )

    attention_mask = instruction["attention_mask"][0] + response["attention_mask"] + [1]
    labels = (
            [-100] * len(instruction["input_ids"][0])
            + response["input_ids"]
            + [tokenizer.pad_token_id]
    )
    if len(input_ids) > MAX_LENGTH:  # 做一个截断
        input_ids = input_ids[:MAX_LENGTH]
        attention_mask = attention_mask[:MAX_LENGTH]
        labels = labels[:MAX_LENGTH]

    input_ids = torch.tensor(input_ids)
    attention_mask = torch.tensor(attention_mask)
    labels = torch.tensor(labels)
    inputs['pixel_values'] = torch.tensor(inputs['pixel_values'])  # float16 ✅ 添加：转换为float16减少内存
    inputs['image_grid_thw'] = torch.tensor(inputs['image_grid_thw']).squeeze(0)  # float16 由（1,h,w)变换为（h,w）
    return {
        "input_ids": input_ids, 
        "attention_mask": attention_mask, 
        "labels": labels,
        "pixel_values": inputs['pixel_values'], 
        "image_grid_thw": inputs['image_grid_thw']
    }


def predict(messages, model):
    # 准备推理
    text = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    image_inputs, video_inputs = process_vision_info(messages)
    inputs = processor(
        text=[text],
        images=image_inputs,
        videos=video_inputs,
        padding=True,
        return_tensors="pt",
    )
    inputs = inputs.to("cuda")

    # 生成输出
    generated_ids = model.generate(**inputs, max_new_tokens=128)
    generated_ids_trimmed = [
        out_ids[len(in_ids) :] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)
    ]
    output_text = processor.batch_decode(
        generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False
    )
    
    return output_text[0]


# 在modelscope上下载Qwen2-VL模型到本地目录下
model_dir = snapshot_download("Qwen/Qwen2-VL-7B-Instruct", cache_dir="./", revision="master")

# 使用Transformers加载模型权重
tokenizer = AutoTokenizer.from_pretrained("./Qwen/Qwen2-VL-7B-Instruct/", use_fast=False, trust_remote_code=True)
processor = AutoProcessor.from_pretrained("./Qwen/Qwen2-VL-7B-Instruct")

model = Qwen2VLForConditionalGeneration.from_pretrained("./Qwen/Qwen2-VL-7B-Instruct/", device_map="auto", torch_dtype=torch.bfloat16, trust_remote_code=True,)
model.enable_input_require_grads()  # 开启梯度检查点时，要执行该方法

# ==================== 在这里添加GPU负载均衡代码 ====================
print(f"检测到 {torch.cuda.device_count()} 个GPU可用")

# 方法1：使用平衡分配策略
if torch.cuda.device_count() > 1:
    print("重新加载模型以实现GPU负载均衡...")
    
    # 方案A：使用balanced策略（推荐）
    try:
        model = Qwen2VLForConditionalGeneration.from_pretrained(
            "./Qwen/Qwen2-VL-7B-Instruct/", 
            device_map="balanced",  # 改为balanced
            torch_dtype=torch.bfloat16, 
            trust_remote_code=True,
        )
        model.enable_input_require_grads()
        print("✅ 使用balanced策略重新加载模型完成")
    except Exception as e:
        print(f"balanced策略失败: {e}")
        print("⚠️  继续使用auto策略")
    
    # 方案B：手动设置显存限制
    max_memory = {i: "14GB" for i in range(torch.cuda.device_count())}
    max_memory["cpu"] = "30GB"
    
    try:
        model = Qwen2VLForConditionalGeneration.from_pretrained(
            "./Qwen/Qwen2-VL-7B-Instruct/", 
            device_map="balanced",
            max_memory=max_memory,  # 添加显存限制
            torch_dtype=torch.bfloat16, 
            trust_remote_code=True,
        )
        model.enable_input_require_grads()
        print("✅ 使用显存限制重新加载模型完成")
    except Exception as e:
        print(f"显存限制加载失败: {e}")
        print("⚠️  保持原有模型")


# 处理数据集：读取json文件
# 拆分成训练集和测试集，保存为data_vl_train.json和data_vl_test.json
train_json_path = "data_vl.json"

with open(train_json_path, 'r') as f:
    data = json.load(f)
    
    # 8:1:1划分（4000:500:500）
    total = len(data)
    train_end = int(total * 0.8)      # 4000
    val_end = train_end + int(total * 0.1)  # 4500
    
    train_data = data[:train_end]     # 训练集
    val_data = data[train_end:val_end]   # 验证集
    test_data = data[val_end:]        # 测试集
    
    print(f"数据划分：训练集 {len(train_data)}，验证集 {len(val_data)}，测试集 {len(test_data)}")

# 保存三个数据集
with open("data_vl_train.json", "w") as f:
    json.dump(train_data, f)

with open("data_vl_val.json", "w") as f:
    json.dump(val_data, f)

with open("data_vl_test.json", "w") as f:
    json.dump(test_data, f)

# 创建训练集和验证集的Dataset
train_ds = Dataset.from_json("data_vl_train.json")
val_ds = Dataset.from_json("data_vl_val.json")

# 使用 remove_columns 移除原始字段
train_dataset = train_ds.map(
    process_func,
    remove_columns=['id', 'conversations']  # 移除这两个原始列
)
val_dataset = val_ds.map(
    process_func,
    remove_columns=['id', 'conversations']  # 同样移除
)

# 验证数据集列
print("处理后训练集列名:", train_dataset.column_names)
print("处理后验证集列名:", val_dataset.column_names)

# 配置LoRA
config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    inference_mode=False,  # 训练模式
    r=64,  # Lora 秩
    lora_alpha=16,  # Lora alaph，具体作用参见 Lora 原理
    lora_dropout=0.05,  # Dropout 比例
    bias="none",
)

# 获取LoRA模型
peft_model = get_peft_model(model, config)

# 配置训练参数
args = TrainingArguments(
    output_dir="./output/Qwen2-VL-7B-optimized",
    # 核心优化参数
    per_device_train_batch_size=12,      # 6比8更安全，效果相近
    gradient_accumulation_steps=3,       # 3比2更稳定
    num_train_epochs=5,                  # 合理
    # 优化器设置
    learning_rate=2e-4,
    warmup_steps=50,
    lr_scheduler_type="cosine",
    # 精度设置（关键！）
    bf16=True,                           # T4支持bf16，优于fp16
    fp16=False,
    # 训练优化
    gradient_checkpointing=False,        # 关闭用现存换速度
    dataloader_num_workers=8,
    dataloader_pin_memory=True,
    dataloader_prefetch_factor=2,        # 新增：预取因子
    # 验证与保存（新增）
    evaluation_strategy="steps",         # 每N步评估
    eval_steps=200,                      # 每200步评估一次
    save_strategy="steps",
    save_steps=400,
    load_best_model_at_end=True,         # 保存最佳模型
    metric_for_best_model="eval_loss",   # 使用验证损失
    greater_is_better=False,
    # 日志
    logging_steps=100,
    report_to="none",
    # 多GPU优化
    ddp_find_unused_parameters=False,
    ddp_timeout=1800,                    # 增加超时时间    
    # remove_unused_columns=True,
)
        
# 设置SwanLab回调
swanlab_callback = SwanLabCallback(
    project="Qwen2-VL-finetune",
    experiment_name="qwen2(7B)-vl-coco2014-500",
    config={
        "model": "https://modelscope.cn/models/Qwen/Qwen2-VL-7B-Instruct",
        "dataset": "https://modelscope.cn/datasets/modelscope/coco_2014_caption/quickstart",
        "github": "https://github.com/datawhalechina/self-llm",
        "prompt": "COCO Yes: ",
        "train_data_number": len(train_data),
        "lora_rank": 64,
        "lora_alpha": 16,
        "lora_dropout": 0.1,
    },
)

# 配置Trainer
trainer = Trainer(
    model=peft_model,
    args=args,
    train_dataset=train_dataset,
    data_collator=DataCollatorForSeq2Seq(tokenizer=tokenizer, padding=True),
    callbacks=[swanlab_callback],
    eval_dataset=val_dataset,           # 添加这行
)


# ==================== 在这里添加GPU监控代码 ====================
print("🔍 启动GPU使用监控...")

# 简单的GPU监控函数
def monitor_gpu_usage(interval=30):
    """监控GPU使用情况"""
    import time
    import pandas as pd
    from datetime import datetime
    
    print(f"GPU监控启动，每{interval}秒记录一次")
    print("-" * 60)
    
    usage_records = []
    try:
        while True:
            record = {"timestamp": datetime.now().strftime("%H:%M:%S")}
            for i in range(torch.cuda.device_count()):
                # 获取显存使用
                mem_alloc = torch.cuda.memory_allocated(i) / 1024**3  # GB
                mem_cached = torch.cuda.memory_reserved(i) / 1024**3  # GB
                util = torch.cuda.utilization(i) if hasattr(torch.cuda, 'utilization') else 0
                
                record[f"gpu{i}_used"] = f"{mem_alloc:.2f}GB"
                record[f"gpu{i}_cached"] = f"{mem_cached:.2f}GB"
                record[f"gpu{i}_util"] = f"{util}%"
            
            # 获取CPU内存使用（Linux系统）
            try:
                import psutil
                cpu_mem = psutil.virtual_memory()
                record["cpu_mem"] = f"{cpu_mem.used/1024**3:.2f}/{cpu_mem.total/1024**3:.2f}GB"
                record["cpu_percent"] = f"{cpu_mem.percent}%"
            except:
                record["cpu_mem"] = "N/A"
                record["cpu_percent"] = "N/A"
            
            # 打印当前状态
            gpu_status = []
            for i in range(torch.cuda.device_count()):
                gpu_status.append(f"GPU{i}: {record[f'gpu{i}_used']}({record[f'gpu{i}_util']})")
            
            print(f"[{record['timestamp']}] {' | '.join(gpu_status)} | CPU内存: {record.get('cpu_mem', 'N/A')}")
            
            usage_records.append(record)
            time.sleep(interval)
            
    except KeyboardInterrupt:
        print("\nGPU监控停止")
        # 保存监控记录
        if usage_records:
            df = pd.DataFrame(usage_records)
            df.to_csv("gpu_monitor_log.csv", index=False)
            print("✅ GPU使用记录已保存到 gpu_monitor_log.csv")

# 启动监控（在后台线程中）
import threading
monitor_thread = threading.Thread(target=monitor_gpu_usage, kwargs={"interval": 30}, daemon=True)
monitor_thread.start()
print("✅ GPU监控已启动（后台运行）")


# 开启模型训练
trainer.train()

# ====================测试模式===================
# 配置测试参数
val_config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    inference_mode=True,  # 训练模式
    r=64,  # Lora 秩
    lora_alpha=16,  # Lora alaph，具体作用参见 Lora 原理
    lora_dropout=0.05,  # Dropout 比例
    bias="none",
)

# 动态查找最新的checkpoint（因为你现在用的是新的输出目录）
import glob
import os

checkpoint_dir = "./output/Qwen2-VL-7B-optimized"

# 查找最新的checkpoint
checkpoints = glob.glob(os.path.join(checkpoint_dir, "checkpoint-*"))
if checkpoints:
    # 按步数排序，取最新的
    checkpoints = [c for c in checkpoints if "checkpoint-best" not in c]
    if checkpoints:
        checkpoints_sorted = sorted(
            checkpoints,
            key=lambda x: int(x.split("-")[-1]) if x.split("-")[-1].isdigit() else 0
        )
        latest_checkpoint = checkpoints_sorted[-1]
        print(f"加载最新模型: {latest_checkpoint}")
        val_peft_model = PeftModel.from_pretrained(model, model_id=latest_checkpoint, config=val_config)
    else:
        print("未找到有效checkpoint，使用原始模型")
        val_peft_model = model
else:
    print("未找到任何checkpoint，使用原始模型")
    val_peft_model = model


# 读取测试数据
with open("data_vl_test.json", "r") as f:
    test_dataset = json.load(f)

test_image_list = []
for item in test_dataset:
    input_image_prompt = item["conversations"][0]["value"]
    # 去掉前后的<|vision_start|>和<|vision_end|>
    origin_image_path = input_image_prompt.split("<|vision_start|>")[1].split("<|vision_end|>")[0]
    
    messages = [{
        "role": "user", 
        "content": [
            { 
            "type": "image", 
            "image": origin_image_path
            },
            {
            "type": "text",
            "text": "COCO Yes:"
            }
        ]}]
    
    response = predict(messages, val_peft_model)
    messages.append({"role": "assistant", "content": f"{response}"})
    print(messages[-1])

    test_image_list.append(swanlab.Image(origin_image_path, caption=response))

swanlab.log({"Prediction": test_image_list})

# 在Jupyter Notebook中运行时要停止SwanLab记录，需要调用swanlab.finish()
swanlab.finish()
```

```python
from transformers import Qwen2VLForConditionalGeneration, AutoProcessor, CLIPModel, CLIPProcessor
from qwen_vl_utils import process_vision_info
from peft import PeftModel, LoraConfig, TaskType
import json
import numpy as np
import torch
from PIL import Image  # 添加这行
import glob

# ==================== 1. 设置环境变量（解决警告） ====================
import os
import glob
os.environ["TOKENIZERS_PARALLELISM"] = "false"

# ==================== 2. 导入专业评估库 ====================
try:
    from pycocoevalcap.cider.cider import Cider
    from pycocoevalcap.spice.spice import Spice
    print("✓ 成功导入 CIDEr 和 SPICE 库")
except ImportError:
    print("正在安装 pycocoevalcap...")
    # Kaggle 环境中通常需要这样安装
    !pip install pycocoevalcap -q
    from pycocoevalcap.cider.cider import Cider
    from pycocoevalcap.spice.spice import Spice

try:
    from bert_score import score as bert_score
    print("✓ 成功导入 BERTScore 库")
except ImportError:
    print("正在安装 bert-score...")
    !pip install bert-score -q
    from bert_score import score as bert_score

# ==================== 3. 加载 CLIP 模型（用于 CLIP-Score）====================
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to("cuda")
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

# ==================== 4. 初始化评估器 ====================
cider_scorer = Cider()
spice_scorer = Spice()

# 读取测试数据
with open("/kaggle/working/data_vl_test.json", "r") as f:
    test_dataset = json.load(f)

# 测试前20张图片
test_count = min(20, len(test_dataset))  # 确保不超过数据集大小
print(f"开始测试前{test_count}张图片...\n")

# 加载LoRA配置
config = LoraConfig(
    task_type=TaskType.CAUSAL_LM,
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
    inference_mode=True,
    r=64,
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
)

# 加载基础模型
base_model = Qwen2VLForConditionalGeneration.from_pretrained(
    "./Qwen/Qwen2-VL-7B-Instruct", 
    torch_dtype="auto", 
    device_map="auto"
)

# 检查模型checkpoint
checkpoint_dir = "./output/Qwen2-VL-7B-optimized"  # ⚠️ 确认路径
best_checkpoint = os.path.join(checkpoint_dir, "checkpoint-best")

if os.path.exists(best_checkpoint):
    print(f"✅ 加载最佳模型: {best_checkpoint}")
    model = PeftModel.from_pretrained(base_model, model_id=best_checkpoint, config=config)
elif os.path.exists(checkpoint_dir):
    checkpoints = glob.glob(os.path.join(checkpoint_dir, "checkpoint-*"))
    if checkpoints:
        checkpoints = [c for c in checkpoints if "checkpoint-best" not in c]
        checkpoints_sorted = sorted(
            checkpoints,
            key=lambda x: int(x.split("-")[-1]) if x.split("-")[-1].isdigit() else 0
        )
        latest = checkpoints_sorted[-1] if checkpoints_sorted else None
        if latest:
            print(f"⚠️  加载最新模型: {latest}")
            model = PeftModel.from_pretrained(base_model, model_id=latest, config=config)
        else:
            print("❌ 无法识别checkpoint，使用基础模型")
            model = base_model
    else:
        print("❌ 没有找到checkpoint，使用基础模型")
        model = base_model
else:
    print(f"❌ 目录不存在: {checkpoint_dir}，使用基础模型")
    model = base_model

# 加载处理器
processor = AutoProcessor.from_pretrained("./Qwen/Qwen2-VL-7B-Instruct")
print("✅ 模型加载完成，开始测试...\n")

all_results = []
ground_truths_all = []
predictions_all = []

for i in range(test_count):
    item = test_dataset[i]
    image_path = item["conversations"][0]['value'].split("<|vision_start|>")[1].split("<|vision_end|>")[0]
    ground_truth = item["conversations"][1]['value']
    
    print(f"\n{'='*60}")
    print(f"测试图片 {i+1}/{test_count}")
    print(f"图像路径: {image_path}")
    print(f"真实描述: {ground_truth}")
    
    # 构建消息
    messages = [
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "image": image_path
                },
                {
                    "type": "text", 
                    "text": "COCO Yes:"
                }
            ]
        }
    ]
    
    # 准备推理
    text = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    image_inputs, video_inputs = process_vision_info(messages)
    inputs = processor(
        text=[text],
        images=image_inputs,
        videos=video_inputs,
        padding=True,
        return_tensors="pt",
    )
    inputs = inputs.to("cuda")
    
    # 生成输出
    generated_ids = model.generate(**inputs, max_new_tokens=128)
    generated_ids_trimmed = [
        out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)
    ]
    output_text = processor.batch_decode(
        generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False
    )
    
    generated_caption = output_text[0]
    print(f"模型生成: {generated_caption}")
    
    # 存储用于批量计算
    ground_truths_all.append(ground_truth)
    predictions_all.append(generated_caption)
    
    # 计算 BERTScore (P, R, F1)
    # 使用默认的 roberta-large 模型，语言设为英文
    P, R, F1 = bert_score([generated_caption], [ground_truth], lang='en', verbose=False)
    
    # 计算 CLIP-Score
    # 准备图像和文本输入
    try:
        image = Image.open(image_path).convert("RGB")
        clip_inputs = clip_processor(text=[generated_caption], images=image, return_tensors="pt", padding=True)
        clip_inputs = {k: v.to("cuda") for k, v in clip_inputs.items()}
        
        # 前向传播
        with torch.no_grad():
            clip_outputs = clip_model(**clip_inputs)
        
        # 计算图像和文本特征的余弦相似度
        logits_per_image = clip_outputs.logits_per_image  # 图像到文本的相似度
        clip_score = logits_per_image.cpu().numpy()[0][0] / 100  # CLIP的logits通常很大，除以100便于比较
    except Exception as e:
        print(f"计算 CLIP-Score 时出错: {e}")
        clip_score = 0.0
    
    # 存储结果（CIDEr和SPICE稍后批量计算）
    result_entry = {
        'image_id': i,
        'image_path': image_path,
        'ground_truth': ground_truth,
        'prediction': generated_caption,
        'scores': {
            'bertscore_precision': float(P[0].cpu().numpy()),
            'bertscore_recall': float(R[0].cpu().numpy()),
            'bertscore_f1': float(F1[0].cpu().numpy()),
            'clip_score': float(clip_score),
        }
    }
    all_results.append(result_entry)
    
    # 打印当前结果
    print(f"图片 {i+1} 部分评估:")
    print(f"  BERTScore-F1: {result_entry['scores']['bertscore_f1']:.4f}")
    print(f"  CLIP-Score:   {result_entry['scores']['clip_score']:.4f}")

# ==================== 5. 批量计算 CIDEr 和 SPICE ====================
print(f"\n{'='*60}")
print("批量计算 CIDEr 和 SPICE...")

# 准备 CIDEr/SPICE 需要的格式
gts = {}
res = {}
for i in range(test_count):
    gts[i] = [ground_truths_all[i]]
    res[i] = [predictions_all[i]]

# 计算 CIDEr（批量）
print("计算 CIDEr...")
cider_score, cider_scores = cider_scorer.compute_score(gts, res)
print(f"CIDEr 计算完成")

# 计算 SPICE（批量）
print("计算 SPICE...（这可能需要一些时间）")
try:
    spice_score, spice_scores = spice_scorer.compute_score(gts, res)
    print(f"SPICE 计算完成")
except Exception as e:
    print(f"SPICE 计算出错: {e}")
    spice_scores = [{"All": {"f": 0.0}} for _ in range(test_count)]

# 将 CIDEr 和 SPICE 结果添加到所有结果中
for i in range(test_count):
    # 处理 CIDEr 分数
    if isinstance(cider_scores, np.ndarray):
        cider_val = float(cider_scores[i])
    elif isinstance(cider_scores, list):
        cider_val = float(cider_scores[i])
    else:
        cider_val = float(cider_scores.get(i, 0.0))
    
    # 处理 SPICE 分数
    if isinstance(spice_scores, list) and i < len(spice_scores):
        if isinstance(spice_scores[i], dict):
            spice_val = spice_scores[i].get("All", {}).get("f", 0.0)
        else:
            spice_val = float(spice_scores[i])
    else:
        spice_val = 0.0
    
    all_results[i]['scores']['cider'] = cider_val
    all_results[i]['scores']['spice'] = spice_val

# ==================== 6. 生成综合报告 ====================
print(f"\n{'='*60}")
print("专业评估综合报告")
print(f"{'='*60}")

# 计算平均值
avg_scores = {
    'CIDEr': np.mean([r['scores']['cider'] for r in all_results]),
    'SPICE': np.mean([r['scores']['spice'] for r in all_results]),
    'BERTScore-F1': np.mean([r['scores']['bertscore_f1'] for r in all_results]),
    'CLIP-Score': np.mean([r['scores']['clip_score'] for r in all_results]),
}

print(f"平均 CIDEr:     {avg_scores['CIDEr']:.4f}")
print(f"平均 SPICE:     {avg_scores['SPICE']:.4f}")
print(f"平均 BERTScore-F1: {avg_scores['BERTScore-F1']:.4f}")
print(f"平均 CLIP-Score:   {avg_scores['CLIP-Score']:.4f}")

# 打印每张图片的详细结果
print(f"\n每张图片的详细结果:")
for i, result in enumerate(all_results):
    print(f"\n图片 {i+1}:")
    print(f"  参考: {result['ground_truth']}")
    print(f"  生成: {result['prediction']}")
    print(f"  CIDEr: {result['scores']['cider']:.4f}, SPICE: {result['scores']['spice']:.4f}, "
          f"BERTScore-F1: {result['scores']['bertscore_f1']:.4f}, CLIP: {result['scores']['clip_score']:.4f}")

# 保存详细结果
with open('professional_evaluation_results.json', 'w') as f:
    json.dump({
        'config': 'Qwen2-VL-7B LoRA微调模型',
        'test_count': test_count,
        'average_scores': avg_scores,
        'detailed_results': all_results
    }, f, indent=2)
print(f"\n详细结果已保存至 'professional_evaluation_results.json'")
```

```python
# 一键创建核心备份
import zipfile
import os
import json

print("🔍 扫描重要文件...")

# 获取重要文件列表
important = []
for item in os.listdir("/kaggle/working"):
    item_path = f"/kaggle/working/{item}"
    if item.endswith(('.json', '.py', '.ipynb', '.txt', '.log')):
        important.append(item_path)
    elif item in ['output', 'Qwen', 'swanlog'] and os.path.isdir(item_path):
        important.append(item_path)

print(f"找到 {len(important)} 个重要项目")

# 创建备份
backup_path = "/kaggle/working/kaggle_backup.zip"
try:
    with zipfile.ZipFile(backup_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for item in important:
            if os.path.isfile(item):
                zipf.write(item, os.path.basename(item))
                print(f"✓ {os.path.basename(item)}")
            elif os.path.isdir(item):
                # 只打包目录中的前20个文件（避免太大）
                file_count = 0
                for root, dirs, files in os.walk(item):
                    for file in files:
                        if file_count < 20:  # 限制每个目录的文件数
                            file_path = os.path.join(root, file)
                            arcname = os.path.join(
                                os.path.basename(item),
                                os.path.relpath(file_path, item)
                            )
                            zipf.write(file_path, arcname)
                            file_count += 1
    
    size_mb = os.path.getsize(backup_path) / 1024 / 1024
    print(f"\n✅ 备份创建成功！")
    print(f"📦 文件: {backup_path}")
    print(f"📏 大小: {size_mb:.2f} MB")
    print(f"\n📥 下载说明：")
    print("1. 在Kaggle左侧边栏点击'Data'")
    print("2. 找到'/kaggle/working'文件夹")
    print("3. 右键点击'kaggle_backup.zip'选择'Download'")
    
except Exception as e:
    print(f"❌ 创建备份失败: {e}")
```
